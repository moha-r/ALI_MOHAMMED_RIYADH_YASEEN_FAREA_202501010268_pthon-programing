# 1. Identify the Components:
1.1. Inputs: quiz_1, quiz_2, quiz_3, choice
1.2. Process: average = (quiz_1 + quiz_2 + quiz_3) / 3. Then, evaluate if average >= 50.
1.3. Output: Print the average mark, and print the status (Pass/Fail).

# 2. Design the Algorithm (Pseudocode):
INITIALIZE choice = "y"
WHILE choice is equal to "y" OR "Y" DO
    READ quiz_1, quiz_2, quiz_3
    COMPUTE average = (quiz_1 + quiz_2 + quiz_3) / 3
    PRINT average
    IF average >= 50 THEN
        PRINT "Pass"
    ELSE
        PRINT "Fail"
    END IF
    READ choice
END WHILE
PRINT "Program Ended"

# 3. Identify the Algorithm Structure:
2.2.1. Sequence: Sequential execution of inputs, calculations, and printing.
2.2.2. Selection: Using if-else to determine Pass or Fail.
2.2.3. Iteration: Using a while loop to repeat the process for another student.
